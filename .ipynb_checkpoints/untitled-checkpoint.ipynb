```python
def no_grad(func):
    def wrapper(*args, **kwargs):
        self = args[0]
        old_training = getattr(self, 'training', True)
        self.training = False
        result = func(*args, **kwargs)
        self.training = old_training
        return result
    return wrapper

class Layer:
    weights: mx.array
    grad: mx.array
    training: bool
    last_X: mx.array
    last_Z: mx.array

    def __init__(self, *args, **kwargs):
        self.__dict__.update(kwargs)
        self.training = True
        self.grad = None
        self.last_X = None
        self.last_Z = None

    def __str__(self) -> str:
        return f"{type(self).__name__}<{self.__dict__}>"

    def __repr__(self) -> str:
        return str(self)

    def __call__(self, X: mx.array) -> mx.array:
        raise NotImplementedError(f"{type(self).__name__} doit implémenter __call__")

    def backward(self, delta: mx.array) -> mx.array:
        raise NotImplementedError(f"{type(self).__name__} doit implémenter backward")

    def update(self, learningRate: float):
        self.weights = self.weights - learningRate * self.grad

    @classmethod
    def Conv2d(cls, dim: int, shape: tuple, activationFunction: Callable = lambda e: e):
        return ConvolutionalLayer.random(dim, shape, activationFunction)
```

```python
class NeuralLayer(Layer):
    weights: mx.array
    biais: mx.array
    func: Callable
    funcPrime: Callable

    def __init__(self, weights_: mx.array, biais_: mx.array, function: Callable) -> None:
        self.func = function
        self.funcPrime = prime(self.func)
        self.weights = weights_
        self.biais = biais_

    
    @classmethod
    def Linear(cls, input: int, output: int, func: Callable):
        return cls(mx.random.uniform(-1, 1, (output, input)), mx.random.uniform(-1, 1, (output, 1)), func)

    def __str__(self) -> str:
        return f"<NeuralLayer {self.dim} -- {self.weights} --- {self.biais} -- {self.func}"

    def __call__(self, x: mx.array) -> Any:
        if self.weights is not None:
            mul = self.weights @ x
            # print(mul)
            # print(x)
            # print(self.biais)
            return mul + self.biais
        return self.func(x) # type: ignore
    
    def isLinear(self) -> bool:
        return self.func is None
    
    @property
    def dim(self):
        if self.isLinear():
            return self.weights.shape # type: ignore
```

```python
class ConvolutionalLayer(Layer):
    kernel: mx.array
    func: Callable
    prime: Callable
    
    def __init__(self, kernel_: mx.array, func_: Callable = lambda e: e):
        super().__init__()
        self.kernel = kernel_
        self.func = func_
        self.prime = prime(func_)
        
    @classmethod
    def random(cls, dim, shape: tuple, activationFunction: Callable = lambda e: e):
        return cls(mx.random.uniform(-1, 1, (dim, ) + shape), activationFunction)


    def __call__(self, X: mx.array) -> mx.array:
        Z = mx.conv2d(X, self.kernel)
        if self.training:
            self.last_X = X
            self.last_Z = Z
        return self.func(Z)

    def backward(self, delta: mx.array) -> mx.array:
        delta_z = delta * self.prime(self.last_Z)
    
        N = self.X.shape[0]
        self.grad = mx.conv2d(
            mx.transpose(self.last_X, (3, 1, 2, 0)),
            mx.transpose(delta_z, (3, 1, 2, 0))
        )
        self.grad = mx.transpose(self.grad, (3, 1, 2, 0)) / N
        
        return mx.conv_transpose2d(delta_z, self.kernel)
```

```python
class PoolingLayer(Layer):
    shape: tuple
    typ: str
    
    def __init__(self, shape_: tuple, typ_: str):
        super().__init__()
        self.shape = shape_
        self.typ = typ_

    def max2d(self, X: mx.array):
        n, m = X.shape
        nn, nm = n // self.shape[0], m // self.shape[1]
        X_np = np.array(X.tolist()) 
        r = np.zeros((nn, nm))
        grad = np.zeros((n, m))
        
        for i in range(nn):
            for j in range(nm):
                bi, bj = self.shape[0] * i, self.shape[1] * j
                patch = X_np[bi:bi+self.shape[0], bj:bj+self.shape[1]]
                max_val = np.max(patch)
                max_pos = np.unravel_index(np.argmax(patch), patch.shape)
                r[i, j] = max_val
                grad[bi + max_pos[0], bj + max_pos[1]] = 1.0
        
        return mx.array(r), grad
    
    def max(self, X: mx.array):
        n, _, _, d = X.shape
        res = []
        self.grad = []
        for i in range(n):
            r = []
            g = []
            for j in range(d):
                x = X[i, :, :, j]
                r_, g_ = self.max2d(x)
                r.append(r_)
                g.append(g_)

            res.append(mx.stack(r, axis=-1))
            self.grad.append(np.stack(g, axis=-1))
        
        self.grad = mx.array(np.stack(self.grad), dtype=mx.float32)
        return mx.stack(res)

    def __call__(self, X: mx.array) -> mx.array:
        if self.typ == "max":
            return self.max(X)
        raise ValueError(f"Type de pooling inconnu : {self.typ}")
    
```
