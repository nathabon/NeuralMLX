```python
import mlx.core as mx
import mlx
import numpy as np
import matplotlib.pyplot as plt

import h5py
import neuralNetwork2 as nn
from collections.abc import Callable

from sklearn.datasets import fetch_openml
import ssl
ssl._create_default_https_context = ssl._create_stdlib_context # type: ignore

# Affiche une image
def show_image(img: mx.array, label=None):
    plt.imshow(img[:, :, 0].tolist(), cmap='gray')
    if label is not None:
        plt.title(f"Label: {mx.argmax(label).item()}")
    plt.axis('off')
    plt.show()

# Affiche une grille d'images
def show_grid(images, labels=None, guessed_labels=None, n=10):
    fig, axes = plt.subplots(1, n, figsize=(15, 2))
    for i in range(n):
        axes[i].imshow(images[i][:, :, 0].tolist(), cmap='gray')
        if labels is not None and guessed_labels is not None:
            axes[i].set_title(f"{mx.argmax(labels[i]).item()} vs {mx.argmax(guessed_labels[i]).item()}")
        elif labels is not None:
            axes[i].set_title(mx.argmax(labels[i]).item())
        axes[i].axis('off')
    plt.show()
```

```python
file_test = "train_full_100_100"
```

```python
nb = 20
mnist = fetch_openml("mnist_784", version=1)
train_X_d = mnist.data[:nb]
train_Y_d = mnist.target[:nb]
train_X_im = mx.array([mx.array(train_X_d.iloc[i].tolist(), dtype=mx.float32).reshape(28, 28, 1) for i in range(len(train_X_d))], dtype=mx.float32)
train_Y_im = [nn.zeros_hot(10, int(nb)) for nb in train_Y_d.tolist()]
train_X = [mx.array(train_X_d.iloc[i].tolist()).reshape(-1, 1) for i in range(len(train_X_d))]
train_Y = [nn.zeros_hot(10, int(nb)) for nb in train_Y_d.tolist()]
```

```python
net = nn.NeuralNetwork.fromFile(file_test, nn.sigmoid)
```

```python
guessed_Y = mx.array([net(x) for x in train_X])
```

```python
show_grid(train_X_im, train_Y_im, guessed_Y, n=nb)
```

# Création de datasets

```python
scales = [0.1, 0.5, 0.8, 1]
output_dir = "datasets/mnist/"
useRandom = False
inline = False
```

```python
mnist = fetch_openml("mnist_784", version=1)
train_X_d = mnist.data
train_Y_d = mnist.target
```

```python
nb = len(train_X_d)
shape = (0, 0)
if inline:
    shape = (784, 1)
else:
    shape = (28, 28, 1)

train_X_im = [mx.array(train_X_d.iloc[i].tolist(), dtype=mx.float32).reshape(shape) for i in range(len(train_X_d))]
train_Y_im = [nn.zeros_hot(10, int(nb)) for nb in train_Y_d.tolist()]
```

```python
for scale in scales:
    newNb = int(nb * scale)
    filename = f"{scale}_{shape[:2]}"
    if useRandom:
        filename += "_enhanced"
    filename += ".h5"
    train_X = train_X_im[:newNb]
    train_Y = train_Y_im[:newNb]

    with h5py.File(output_dir + filename, "w") as f:
        dsetX = f.create_dataset("train_X", data=train_X)
        dsetY = f.create_dataset("train_Y", data=train_Y)
```
